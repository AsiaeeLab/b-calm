# Reproducing the B-CALM experiments

**B-CALM: Bias-Limited Bayesian Borrowing for RCT-Anchored Treatment Effects under Covariate Mismatch**

This notebook reproduces **all synthetic and semi-synthetic results** in the paper: the saturation diagnostic, the main simulation grid, the calibration study, the STAR-like semi-synthetic benchmark, and the ablations, together with every figure and LaTeX table derived from them.

**What is not reproducible here.** The real-data pediatric-obesity study uses patient-level EHR/trial records that cannot be released for IRB/DUA reasons, so the real-data analysis code and its inputs are excluded from this package.

**Baselines that need a companion repo.** The OSCAR/CALM-family baselines (R-OSCAR, MR-OSCAR, CALM) are prior works whose code lives in a separate companion repository. They are *skipped automatically* unless you check that repo out locally and point the environment variable `BCALM_EXTERNAL_CALM_PATH` at it. B-CALM, RCT-only, CF, pooled, OS-only, and the ablation variants run with just the dependencies in `requirements.txt`.

**Runtime.** The cells below use `--mode full`, which reproduces the paper numbers but takes a while. For a quick end-to-end sanity check, change `--mode full` to `--mode smoke` (tiny, seconds) or `--mode cheap` (intermediate) in each `run_*.py` cell below.
